```python
import pandas as pd

# Excel dosyasını oku 2. csv
file_path = '/content/drive/My Drive/zararlı.xlsx'
df = pd.read_excel(file_path)

# 'Code ID' ve 'Name(technique_description)' sütunlarını seç
df_selected = df[['Code ID', 'Name(technique_description)']]

# 'Code ID' ve 'Name' sütunlarını kullanarak bir pivot table oluştur
pivot_table = pd.pivot_table(df_selected, index='Code ID', columns='Name(technique_description)', aggfunc=len, fill_value=0)

# Her bir hücredeki değerleri 1 veya 0 olarak güncelle
pivot_table = pivot_table.applymap(lambda x: 1 if x > 0 else 0)

# Yeni bir Excel dosyasına kaydet
output_path = '/content/drive/My Drive/öznitelik/technique_description-zararlı.xlsx'
pivot_table.to_excel(output_path)

print("Dosya başarıyla kaydedildi:", output_path)
```

```python
import pandas as pd

# Veri setini yükle
file_path = '/content/drive/My Drive/zararsız.xlsx'
df = pd.read_excel(file_path)

# NaN olmayan benzersiz ATT&CK ID değerlerini al ve string'e çevir
attack_ids = df['ATT&CK ID'].dropna().unique()
attack_ids = [str(aid) for aid in attack_ids]

# Yeni sütun başlıkları oluştur
new_columns = ['m.' + aid for aid in attack_ids] + ['s.' + aid for aid in attack_ids] + ['ı.' + aid for aid in attack_ids]

# Yeni DataFrame oluştur
new_df = pd.DataFrame(columns=['Code ID', 'Code'] + new_columns)

# Code ID'ye göre gruplandır ve her bir grup için işlem yap
for code_id, group in df.groupby('Code ID'):
    row = {'Code ID': code_id, 'Code': group['Code'].iloc[0]}  # Code ID ve Code bilgisini al
    for aid in attack_ids:
        # Her bir ATT&CK ID için m, s, ve ı değerlerini hesapla
        # Burada, herhangi bir değerin var olup olmadığını kontrol ediyoruz
        row['m.' + aid] = 1 if any(group[group['ATT&CK ID'] == aid]['Malicious Indicators'].notna()) else 0
        row['s.' + aid] = 1 if any(group[group['ATT&CK ID'] == aid]['Suspicious Indicators'].notna()) else 0
        row['ı.' + aid] = 1 if any(group[group['ATT&CK ID'] == aid]['Informative Indicators'].notna()) else 0
    new_df = pd.concat([new_df, pd.DataFrame(row, index=[0])], ignore_index=True)

# Yeni DataFrame'i bir Excel dosyasına kaydet
output_path = '/content/drive/My Drive/öznitelik/zararsız_atack_id.xlsx'
new_df.to_excel(output_path, index=False)

print("Dosya başarıyla kaydedildi:", output_path)
```
